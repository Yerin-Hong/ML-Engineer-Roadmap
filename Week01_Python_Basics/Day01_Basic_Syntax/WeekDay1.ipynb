```python
scores = [85, 92, 78, 64, 100, 55, 89]

# 전체 점수 평균
avg_scores = sum(scores) / len(scores)

# 리스트 초기화 먼저
over_than_80 = []
less_than_60 = []

# 80점 이상 학생, 60점 미만 학생
for s in scores:
  if s >= 80:
    over_than_80.append(s)
  elif s < 60:
    less_than_60.append(s)

# 출력
print("전체학생 점수 평균: ", avg_scores)
print("80점 이상 학생의 점수: ", over_than_80)
print("60점 미만 학생의 점수: ", less_than_60)
print("가장 높은 점수: ", max(scores))
print("가장 낮은 점수: ", min(scores))
```

```python
scores = {
    "Kim": 85,
    "Lee": 92,
    "Park": 78,
    "Choi": 65,
    "Jung": 100
}

score_values = []
over_than_80 = []

# 점수만 담긴 리스트 생성
for value in scores.values():
    score_values.append(value)

# 평균 점수
avg_scores = sum(score_values) / len(score_values)

# 80점 이상인 학생의 이름만 리스트에 저장
for key, value in scores.items():
     if value >= 90:
         over_than_80.append(key)

print("평균 점수: ", avg_scores)
print("80점 이상인 학생의 이름: ", over_than_80)
```

```python
# 학생 점수 분석 함수
def calculate_total(math, english, science):
  total = (math + english + science)
  return total

total_score = calculate_total(85, 90, 95)
print(total_score)
```

```python
# 성적 분석 함수에 기본값 매개변수 적용.
# 기본 코드는 위 코드 그대로 불러와서 사용.

# 기본값이 있는 매개변수는 기본값이 없는 매개변수 뒤에 배치해야 함.
def calculate_average(math, english, science=0):
  avg_score = (math + english + science) / 3
  return avg_score

result1 = calculate_average(80, 90, 100)
result2 = calculate_average(80, 90)
print(result1)
print(result2)
```

```python
# *args 로 매개변수의 개수를 미리 정할 수 없거나, 가변 인자로 둘 때 사용.
def calculate_average(*scores):
  total = sum(scores)
  average = total / len(scores)
  return average

print(calculate_average(80, 90, 100))
print(calculate_average(80, 90, 100, 70, 60))
```

```python
# *args 구현
def calculate_total(*scores):
  total = sum(scores)
  return total

print(calculate_total(10, 20, 30))
print(calculate_total(10, 20, 30, 40, 50))
```

```python
# Day 1 종합 미니 프로젝트
students = {
    "Kim": [85, 90, 95],
    "Lee": [70, 80, 75],
    "Park": [95, 100, 90],
    "Choi": [55, 60, 65]
}

# 1. 학생별 평균 점수 계산(함수를 불러와서 계속 쓸 거니까 얜 for문 필요 X. 한 번의 정의만 필요)
def calculate_average(scores):
  stu_avg = sum(scores) / len(scores)
  return stu_avg

# 2. 모든 학생을 순회하면서 평균 출력(역할: 여러 학생 관리)

# 학생 이름과 평균 담을 딕셔너리 생성
student_averages = {}

for name, scores in students.items():
  # 실제 변수
  avg_score = calculate_average(scores)
  # 여기에 학생 이름과 평균 출력
  # 학생 이름, 평균을 딕셔너리에 저장
  student_averages[name] = avg_score
  print(f"{name}의 평균: {avg_score}")
print("==================================================")

# 3. 평균 점수가 80점 이상인 학생의 이름만 over_than_80 리스트에 저장하기.
over_than_80 = []
for name, avg_scores in student_averages.items():
  if avg_scores >= 80:
    over_than_80.append(name)
print("평균 점수가 80점 이상인 학생: ", over_than_80)
print("==================================================")

# 4. 학생 전체 평균(각 학생의 평균을 구하는 코드를 다시 불러와서 사용-각 학생의 평균을 먼저 구하고 그걸 누적합)
# 전체평균 변수 초기화
total_average = 0

# 각 학생의 평균은 student_averages 딕셔너리 이용
total_average = sum(student_averages.values()) / len(student_averages)

print("전체 학생의 평균 점수: ", total_average)
print("==================================================")

# 5. 학생 이름 검색 및 예외 처리
try:
  stu_name = input("학생 이름을 입력하세요: ")
  avg_score = student_averages[stu_name]   # 딕셔너리에서 []로 불러온 값은 key와 대응되는 값임. 즉, name하면 name의 avg_score를 가져오는거.
  print(f"{stu_name}의 평균 점수는 {avg_score}점입니다.")
except KeyError:
  print("해당 학생을 찾을 수 없습니다.")
```
